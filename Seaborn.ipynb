```python
# despine - remove top and right spines
sns.set_style('ticks')
sns.scatterplot(data = df, x = 'flipper_length_mm', y = 'body_mass_g', 
                hue = 'island')
sns.despine()
```

```python
# Set_Context - paper, notebook, talk, poster
sns.set_style('ticks')
sns.set_context('paper')
sns.scatterplot(data = df, x = 'flipper_length_mm', y = 'body_mass_g', 
                hue = 'island')
```

```python
# palette - deep, muted, bright, pastel, dark, colorblind
sns.set_style('ticks')
sns.set_context('paper')
sns.scatterplot(data = df, x = 'flipper_length_mm', 
        y = 'body_mass_g', hue = 'island', palette='pastel')
```

```python
# pair grid - with map functions
pg = sns.PairGrid(data = df, hue = 'sex')
pg.map_upper(sns.scatterplot) # upper triangle
pg.map_lower(sns.kdeplot) # lower triangle
pg.map_diag(sns.histplot) # diagonal
pg.add_legend()
```
