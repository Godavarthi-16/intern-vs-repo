```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 100)
x
```

```python
y = np.sin(x)
y
```

```python
plt.plot(x,y , color='yellow')
plt.title("Sine Wave")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.show()
```

```python
plt.plot(x,y , color='#27E0F5')
plt.title("Cos Wave")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.show()
```

```python
x = np.linspace(-10,10,20)
y = x**2
plt.plot(x,y , color='#9127F5')
plt.title("Parabola")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.show()
```

```python
plt.plot(x,y , 'r+')
plt.title("Parabola")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.show()
```

```python
x = np.linspace(-5,5,50)
plt.plot(x, np.sin(x), 'm--')
plt.plot(x, np.cos(x), 'b-.')
plt.title("Sine and Cosine waves")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend({'Sine','Cosine'})
plt.grid(True)
plt.show()
```

```python
x = np.linspace(-5,5,50)
plt.plot(x, np.sin(x), color='#F52738', linestyle = 'dashdot', linewidth = 2)
plt.plot(x, np.cos(x), color='#E4F527', linestyle = 'dotted', linewidth = 2)
plt.title("Sine and Cosine waves")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend({'Sine','Cosine'})
plt.grid(True)
plt.show()
```

```python
categories = ['A','B','C','D','E','F']
values = [10,15,7,12,4,9]
plt.bar(categories, values, color= '#3527F5')
plt.title("Bar Chart")
plt.xlabel("categories")
plt.ylabel("values")
plt.show()
```

```python
children = [10,20,40,10,30]
colors = ['red red red red red red',
          'blue blue blue blue', 'green green green green green',
          'yellow yellow yellow yellow', 'pink pink pink pink']
plt.barh(colors, children, color = "#27F542")
plt.show()
```

```python
categories = ['A','B','C','D','E']
values = [10,15,7,12,4]
plt.pie(values, labels=categories)
plt.title("Pie chart")
plt.show()
```

```python
categories = ['A','B','C','D','E']
values = [10,15,7,12,4]
plt.pie(values, labels=categories , autopct= '%0.f%%')
plt.title("Pie chart")
plt.show()
```

```python
# Histogram
x = [10, 20, 22, 24, 28, 30, 32, 35, 38, 34, 36, 40, 47, 50]
plt.hist(x, bins=[10, 20, 30, 40, 50], color="#3FDE0E")
plt.title("Histogram Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# Box Plot
# Use case: Identify outliers, visualize distribution, compare datasets
data = [np.random.randn(10) for i in range(4)]
print(data)
plt.boxplot(data)
plt.title("Box Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# Area chart
# Alpha parameter controls transparency - 0 to 1
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.fill_between(x, y1, color='#FFA07A', alpha=1)
plt.fill_between(x, y2, color='#20B2AA', alpha=0.5)
plt.title("Area Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend(['Sine', 'Cosine'])
plt.show() 
```

```python
# Subplot(nrow, ncol, place)

x = np.array([1, 2, 3, 4, 5])
y = x **2
plt.subplot(1, 2, 1)
plt.plot(x, y, color = 'red')

z = x **3
plt.subplot(1, 2, 2)
plt.plot(x, z, color = 'blue')

plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y = x ** 2
plt.subplot(2, 2, 1)
plt.plot(x, y, color = 'red')

z = x ** 3
plt.subplot(2, 2, 2,)
plt.plot(x, z, color = 'blue')

z = x ** 6
plt.subplot(2, 2, 3)
plt.plot(x, z, color = 'cyan')

z = x ** 9
plt.subplot(2, 2, 4)
plt.plot(x, z, color = 'green')

plt.tight_layout()
plt.show()
```

```python
### Figure and figsize(width, height)
ax = plt.figure(figsize = (12, 5))

x = np.array([1, 2, 3, 4, 5])
y = x ** 2
plt.plot(x, y, color = 'red')
plt.show()
```
