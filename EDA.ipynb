```python
df['Family_Size'] = df['SibSp'] + df['Parch'] + 1
df.head()
```

```python
df['Family_Size'].value_counts()
```

```python
# family type
# 0 -> alone
# 2-4 -> small
# 5 -> large
```

```python
def transform_family_size(size):
    if size == 1:
        return 'Alone'
    elif size > 1 and size <= 4:
        return 'Small'
    else:
        return 'Large'
        
```

```python
df['Family_Size'].apply(transform_family_size)
```

```python
df['Family_Size_category'] = df['Family_Size'].apply(transform_family_size)
df.head()
```
