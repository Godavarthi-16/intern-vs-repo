```python
from bs4 import BeautifulSoup
import requests
```

```python
url = "https://companiesmarketcap.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
print(soup.title.string)
```

```python
print(soup.title)
```

```python
soup.find('table')
```

```python
soup.find_all('table')
```

```python
soup.find('table',class_ = "default-table table marketcap-table dataTable")
```

```python
soup.find_all('th')
```

```python
word_titles = soup.find_all('th')
word_titles
```

```python
word_titles = [title.text.strip() for title in word_titles]
word_titles
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns = word_titles)
df
```

```python
column_data = soup.find_all('tr')
column_data
```

```python
for row in column_data:
    print(row.find_all('td'))
```

```python
for row in column_data:
    row_data = row.find_all('td')
    print(row_data)
```

```python
for row in column_data:
    row_data = row.find_all('td')
    print(row_data)
    individual_data = [data.text.strip() for data in row_data]
    print(individual_data)
```

```python
df.columns
```

```python
for row in column_data:
    row_data = row.find_all('td')
    individual_data = [data.text.strip() for data in row_data]

    if len(individual_data) == len(df.columns):
        df.loc[len(df)] = individual_data
```
